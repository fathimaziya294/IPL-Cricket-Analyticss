```python
import sqlite3
import pandas as pd
```

```python
con=sqlite3.connect("../data/raw/ipl.db")
def query(sql):
    return pd.read_sql(sql,con)
```

```python
query('SELECT COUNT(*) AS total_matches FROM matches;')
```

```python
query('''
SELECT DISTINCT season
FROM matches;
''')
```

```python
query('''
SELECT DISTINCT match_id
FROM deliveries;
''')
```

```python
query('''
SELECT DISTINCT bowler_type
FROM deliveries;
''')
```

```python
query('''
SELECT match_id,season,match_winner
FROM matches
WHERE match_winner ='Mumbai Indians';
''')
```

```python
query('''
SELECT match_id,match_date,venue,match_winner
FROM matches
WHERE season ='2018';
''')
```

```python
query('''
SELECT match_id,season,venue,city
FROM matches
WHERE city ='Mumbai';
''')
```

```python
query('''
SELECT match_id,season,venue,match_winner,win_by_runs
FROM matches
WHERE win_by_runs>=150;
''')
```

```python
query('''
SELECT match_id,season,venue,team1,team2
FROM matches
WHERE match_winner IS NULL;
''')
```

```python
query('''
SELECT match_id,venue,match_winner
FROM matches
ORDER BY match_date DESC
LIMIT 10;
''')
```

```python
query('''
SELECT DISTINCT d.batter
FROM deliveries AS d 
LEFT JOIN players p
ON p.player_name =d.batter
WHERE player_name IS NULL
''')
```
