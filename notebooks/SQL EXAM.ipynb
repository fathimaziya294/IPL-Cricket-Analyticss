```python
import pandas as pd 
import sqlite3
con=sqlite3.connect('../data/raw/ipl.db')
def q(sql):
    return pd.read_sql(sql,con)
```

```python
q("""
  SELECT COUNT(*) AS total_deliveries
  FROM deliveries;""")
```

HERE WE USED SELECT and COUNT(*) TO GET THE TOTAL BOWLS.THIS QUERY DESCRIBES THE TOTAL BALLS BOWLED OVER THE PLAYERS IN  TEAMS

```python
q("""
  SELECT match_id,win_by_runs
  FROM matches
  ORDER BY win_by_runs DESC 
  LIMIT 5;""")
```

HERE WE USED DESC IT SHOWS THE HIGHEST RUN VICTORIES ACROSS MATCHES

```python
q("""
  SELECT match_winner, COUNT(*) AS match_won
  FROM matches
  GROUP BY match_winner
  ORDER BY match_won DESC;
  """)
```
